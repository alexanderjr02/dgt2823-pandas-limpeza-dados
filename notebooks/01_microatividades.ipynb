Leitura, visualização e inspeção de um conjunto de dados CSV com a biblioteca Pandas.

Conjunto de dados: `dados/dados.csv` (colunas separadas por `;`).

## Microatividade 1: ler um arquivo CSV

Importa a biblioteca, lê o CSV informando o separador, a engine `python` e o encoding, atribui o resultado a uma variável e exibe o conteúdo.

```python
import pandas as pd

CAMINHO_CSV = '../dados/dados.csv'

df = pd.read_csv(CAMINHO_CSV, sep=';', engine='python', encoding='utf-8')

print(df)
```

## Microatividade 2: criar um subconjunto de dados

Nova variável contendo apenas 3 colunas do conjunto original.

```python
subconjunto = df[['Duration', 'Pulse', 'Calories']]

print(subconjunto)
```

## Microatividade 3: configurar o número máximo de linhas exibidas

Define a opção `display.max_rows` como 9999 e imprime o conjunto original com `to_string()`.

```python
pd.set_option('display.max_rows', 9999)
print('display.max_rows =', pd.get_option('display.max_rows'))

print(df.to_string())
```

## Microatividade 4: exibir as primeiras e as últimas N linhas

```python
print('Primeiras 10 linhas:')
print(df.head(10))
```

```python
print('Últimas 10 linhas:')
print(df.tail(10))
```

## Microatividade 5: informações gerais sobre colunas, linhas e dados

```python
df.info()
```

Os mesmos dados, extraídos um a um:

```python
linhas, colunas = df.shape
print('Total de linhas:', linhas)
print('Total de colunas:', colunas)

print('\nQuantidade de dados nulos por coluna:')
print(df.isnull().sum())

print('\nTipo de dado de cada coluna:')
print(df.dtypes)

memoria = df.memory_usage(deep=False).sum()
print(f'\nMemória utilizada: {memoria} bytes ({memoria / 1024:.1f} KB)')
```

### Respostas (a partir do `df.info()`)

| Informação | Valor |

| Total de linhas | 32 (índices 0 a 31, `RangeIndex: 32 entries`) |
| Total de colunas | 6 |
| Dados nulos | 3 no total: 1 em `Date` (31 não nulos) e 2 em `Calories` (30 não nulos) |
| Tipos de dado | `ID`, `Duration`, `Pulse`, `Maxpulse`: `int64`; `Calories`: `float64` (por causa dos `NaN`); `Date`: texto (`str`/`object`) |
