Problemas conhecidos no conjunto:

| Linha | Coluna | Problema |
|---|---|---|
| 18 | Calories | valor nulo |
| 28 | Calories | valor nulo |
| 22 | Date | valor nulo |
| 26 | Date | data fora do padrão (`20201226`) |

## 1. Leitura do CSV

O arquivo usa `;` como separador, por isso `sep=';'`. A engine `python` e o encoding `utf-8` também são informados.

```python
import numpy as np
import pandas as pd

pd.set_option('display.max_rows', 9999)

CAMINHO_CSV = '../dados/dados.csv'

dados_originais = pd.read_csv(CAMINHO_CSV, sep=';', engine='python', encoding='utf-8')
```

## 2. Verificação da importação

### Informações gerais

```python
dados_originais.info()
```

### Primeiras e últimas N linhas

```python
N = 5

print(f'Primeiras {N} linhas:')
print(dados_originais.head(N))
```

```python
print(f'Últimas {N} linhas:')
print(dados_originais.tail(N))
```

## 3. Cópia do conjunto original

Todas as transformações são feitas na cópia, preservando os dados originais.

```python
dados = dados_originais.copy()
```

## 4. Valores nulos da coluna `Calories` substituídos por 0

```python
dados['Calories'] = dados['Calories'].fillna(0)

print(dados.to_string())
```

Linhas 18 e 28 agora têm `Calories = 0.0`.

## 5. Valores nulos da coluna `Date` substituídos por `1900/01/01`

```python
dados['Date'] = dados['Date'].fillna('1900/01/01')

print(dados.to_string())
```

Linha 22 agora tem `Date = 1900/01/01`.

## 6. Primeira tentativa de converter `Date` para datetime

As datas do arquivo vêm entre aspas simples (`'2020/12/01'`), então o Pandas deduz o formato `'%Y/%m/%d'` a partir do primeiro valor. O `1900/01/01` inserido no passo anterior não tem as aspas e a conversão falha. O `try/except` exibe o erro sem interromper o notebook.

```python
try:
    pd.to_datetime(dados['Date'])
except ValueError as erro:
    print('Erro:', erro)
```

## 7. Correção: `1900/01/01` substituído por `NaN` e nova conversão

```python
dados['Date'] = dados['Date'].replace('1900/01/01', np.nan)

print(dados.to_string())
```

```python
try:
    pd.to_datetime(dados['Date'])
except ValueError as erro:
    print('Erro:', erro)
```

Agora o erro é outro: o valor `20201226` (linha 26) não segue o formato `'%Y/%m/%d'`.

## 8. Correção do valor `20201226` combinando `replace` e `to_datetime`

O valor é uma string sem separadores. Ele é convertido sozinho com `to_datetime` no formato `%Y%m%d` e o resultado entra no lugar da string com `replace`.

```python
dados['Date'] = dados['Date'].replace(
    '20201226', pd.to_datetime('20201226', format='%Y%m%d')
)

print('Linha 26:', dados.loc[26, 'Date'])
```

## 9. Conversão de toda a coluna `Date` para datetime

O formato informado inclui as aspas simples que existem no arquivo. O valor da linha 26, que já é datetime, é mantido; o `NaN` da linha 22 vira `NaT`.

```python
dados['Date'] = pd.to_datetime(dados['Date'], format="'%Y/%m/%d'")

print(dados.to_string())
```

```python
print(dados.dtypes)
```

## 10. Remoção dos registros com valores nulos

Só a linha 22 (coluna `Date`) ainda tem valor nulo, por isso a coluna é usada como base na remoção.

```python
print('Nulos antes da remoção:')
print(dados.isnull().sum())

dados = dados.dropna(subset=['Date'])
```

## 11. Resultado final

```python
print(dados.to_string())
```

```python
dados.info()
```

Conferência das transformações:

- `Calories`: linhas 18 e 28 com 0, sem nulos;
- `Date`: tipo `datetime64`, linha 26 convertida para `2020-12-26`;
- linha 22 removida, restando 31 registros sem nenhum valor nulo.
